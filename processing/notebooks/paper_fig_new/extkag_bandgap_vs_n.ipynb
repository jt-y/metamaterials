```python
import os
import pickle as pickle
import glob
import skimage
import skimage.transform
import numpy as np
from matplotlib import pyplot as plt
from tqdm import tqdm
import csv

import sys
```

```python
data_dir = "../../../data/simulation/"
out_dir = "../../../output/paper_new/extkag"
```

```python
cavity_num = []
q = []
gap = []
bandwidth = []

with open(data_dir+'extkag.csv', newline='') as csvfile:
    reader = csv.reader(csvfile, delimiter=',')
    next(reader)
    
    for row in reader:
        cavity_num.append(int(row[0]))
        q.append(float(row[1]))
        gap.append(float(row[2]))
        bandwidth.append(float(row[3]))
        
```

```python
fig, ax = plt.subplots(figsize = (2, 1.5), tight_layout = True)
ax.plot(np.array(cavity_num)[0:5], np.array(gap)[0:5], 'k:', linewidth = 0.5)
ax.plot(np.array(cavity_num)[0:5], np.array(gap)[0:5], 'b.', markersize = 5)

for axis in ['top','bottom','left','right']:
    ax.spines[axis].set_linewidth(0.25)

plt.xticks(np.linspace(0, 4, 5))
plt.yticks(np.linspace(0, 800, 3))

ax.xaxis.set_tick_params(width=0.25, length = 1)
ax.yaxis.set_tick_params(width=0.25, length = 1)

ax.tick_params(labelbottom=False)
ax.tick_params(labelleft=False)
plt.show()
fig.savefig(out_dir+"/extkag_bandgap.pdf", bbox_inches = 'tight')
```

```python
fig, ax = plt.subplots(figsize = (2, 1.5), tight_layout = True)

ax.plot(np.array(cavity_num)[0:5], np.array(bandwidth)[0:5], 'k:', linewidth = 0.5)
ax.plot(np.array(cavity_num)[0:5], np.array(bandwidth)[0:5], 'b.', markersize = 5)
ax.set_yscale('log')

plt.xticks(np.linspace(0, 4, 5))
plt.yticks(10**np.linspace(-4, 2, 3))


ax.xaxis.set_tick_params(width=0.25, length = 1)
ax.yaxis.set_tick_params(width=0.25, length = 1)
for axis in ['top','bottom','left','right']:
    ax.spines[axis].set_linewidth(0.25)

ax.tick_params(labelbottom=False)
ax.tick_params(labelleft=False)
plt.show()
fig.savefig(out_dir+"/extkag_bandwidth.pdf", bbox_inches = 'tight')

```
